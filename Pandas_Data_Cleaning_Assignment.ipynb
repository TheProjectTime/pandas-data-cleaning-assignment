**Step 1:** Importing Pandas

```python
import pandas as pd
```

**Step 2:** Loading the CSV

```python
df = pd.read_csv("data.csv")

print(df)
```

**Step 3:** Loading the first five rows were displayed to understand the structure of the data.

```python
print(df.head())
```

**Step 4:** Checking dataset information

```python
print("Shape of dataset:", df.shape)
print("\nDataset information:")
df.info()
```

**Step 5:** Identifying Missing Values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

**Step 6:** Handling Missing Values

Missing values in the Calories column were replaced with the mean Calories value. The missing Date value was replaced with "Unknown".

For the Calories column:

```python
df['Calories'] = df['Calories'].fillna(df['Calories'].mean())
```

For the missing Date, we'll fill it with "Unknown":

```python
df['Date'] = df['Date'].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

**Step 7:** Removing Duplicate Rows

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

**Step 8:** Filtering Rows

Filtering workouts where Pulse is greater than 100:

```python
high_pulse = df[df['Pulse'] > 100]

print(high_pulse)
```

Filtering workouts where Pulse is greater than 300:

```python
high_calories = df[df['Calories'] > 300]

print(high_calories)
```

**Step 9:** Creating a New Column

A new column called **Calories_per_Minute** was created by dividing Calories by Duration.

```python
df['Calories_per_Minute'] = df['Calories'] / df['Duration']

print(df.head())
```

**Step 10:** Creating Workout Categories

A new **Workout_Type** column was created to classify workouts as Long or Short based on their duration.

```python
df['Workout_Type'] = df['Duration'].apply(
    lambda x: 'Long' if x >= 60 else 'Short'
)

print(df.head())
```

**Step 11:** Displaying the final cleaned dataset

```python
print("Final Cleaned Dataset:")
print(df)
```

**Step 12:** Saving the cleaned CSV

```python
df.to_csv("cleaned_data.csv", index=False)

print("Cleaned dataset saved successfully.")
```

**Conclusion**

The dataset was successfully cleaned using Pandas. Missing values were handled, duplicate records were removed, rows were filtered based on conditions, and new columns were created. The final cleaned dataset was then saved as a CSV file.
